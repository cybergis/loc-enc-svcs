# GeoShapley Decomposition Visualizations

Creates simple bar charts showing:
1. **GeoShapley decomposition**: Primary (X1, X2) + Geographic + Interactions
2. **GeoShapley with embeddings**: Primary + Embeddings  
3. **Standard SHAP**: Simple feature contributions

## Setup

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
plt.rcParams['figure.dpi'] = 150
plt.rcParams['font.size'] = 14
```

## Figure 1: GeoShapley Decomposition (Geographic)

```python
# Simulated GeoShapley decomposition for one observation
# Based on GeoShapley paper: Primary + Geographic + Interaction components

components = ['X1\n(Primary)', 'X2\n(Primary)', 'X1 × Location\n(Interaction)', 
              'X2 × Location\n(Interaction)', 'Geographic']
values = [2.3, -1.5, 1.8, -0.9, 2.1]  # Example SHAP contributions

colors = ['#4ecdc4' if v > 0 else '#ff6b6b' for v in values]

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(components, values, color=colors, edgecolor='black', linewidth=2)
ax.axvline(0, color='black', linewidth=2)
ax.set_xlabel('Contribution to Prediction')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('fig1_geoshapley_geographic.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```

## Figure 2: GeoShapley with Embeddings

```python
# With location embeddings instead of raw coordinates
components_emb = ['X1\n(Primary)', 'X2\n(Primary)', 'X1 × Emb\n(Interaction)', 
                  'X2 × Emb\n(Interaction)', 'emb_0', 'emb_1', 'emb_2', 'emb_3']
values_emb = [2.1, -1.4, 1.2, -0.7, 1.5, 0.9, -1.1, 0.6]

colors_emb = ['#4ecdc4' if v > 0 else '#ff6b6b' for v in values_emb]

fig, ax = plt.subplots(figsize=(10, 7))
bars = ax.barh(components_emb, values_emb, color=colors_emb, edgecolor='black', linewidth=2)
ax.axvline(0, color='black', linewidth=2)
ax.set_xlabel('Contribution to Prediction')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('fig2_geoshapley_embeddings.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```

## Figure 3: Standard SHAP (Non-spatial)

```python
# Standard SHAP - no spatial decomposition, just feature contributions
features = ['Age', 'Income', 'Education', 'Experience']
values_shap = [3.2, 5.1, -2.3, 1.4]

colors_shap = ['#4ecdc4' if v > 0 else '#ff6b6b' for v in values_shap]

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.barh(features, values_shap, color=colors_shap, edgecolor='black', linewidth=2)
ax.axvline(0, color='black', linewidth=2)
ax.set_xlabel('Contribution to Prediction')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('fig3_shap_standard.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```
